```python
import matplotlib.pyplot as plt
import numpy as np
import math
```

## Task I

```python
x_points = np.linspace(start = -10, stop = 10, num = 200, endpoint = True)

y1 = x_points
y2 = x_points ** 2
y3 = np.sin(x_points)
y4 = np.exp(-0.1 * x_points) * np.cos(x_points)

plt.plot(x_points, y1, c = "r", label = "y = x")
plt.plot(x_points, y2, c = "g", label = "y = x^2")
plt.plot(x_points, y3, c = "b", label = "y = sin(x)")
plt.plot(x_points, y4, c = "y", label = "y = e^(0.1x)*cos(x)")
plt.xlabel("X Data")
plt.ylabel("Y Data")
plt.legend()
plt.grid()
plt.title("Simple Graphs")
```

## Task 2

```python
t = np.linspace(0, 2 * np.pi, 500)

# this equation in its parametric form represents the heart equation using basic trignometric functions
x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

plt.plot(x, y, color="r")
plt.grid()
plt.xlabel("X Data")
plt.ylabel("Y Data")
plt.title("Parametric Curve For Heart Equation")
```

## Task 3

```python
students = ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"]
midterm = [85, 72, 90, 66, 78, 92, 60, 74, 88, 95]
final = [80, 70, 94, 68, 75, 90, 65, 72, 84, 96]

def total(m, f):
    return 0.4 * m  + 0.6 * f

aggregate = list(map(total, midterm, final)) # total scores

plt.scatter(midterm, final, c = "b")
plt.xlabel("Midterm Scores")
plt.ylabel("Final Scores")
plt.title("Scores Scatter Plot")
plt.grid()
```

```python
plt.hist(aggregate, bins=10, color="skyblue", edgecolor="black")
plt.title("Distribution of Total Scores")
```

```python
plt.bar(students, aggregate, edgecolor="black")
plt.title("Individual Score Distribution")
```

## Task 4

```python
# plotting the best fit line
slope, intercept = np.polyfit(midterm, final, 1)
midterm = np.array(midterm)
final = np.array(final)
fit_line = slope * midterm + intercept

plt.scatter(midterm, final, c = "r", label = "Data Points")
plt.plot(midterm, fit_line, c = "b", label = "Best Fit Line")
plt.xlabel("Midterm Scores")
plt.ylabel("Final Scores")
plt.title("Scores Scatter Plot")
plt.legend()
plt.grid()

print(fit_line)
```
